```python
# ═══════════════════════════════════════════════════════════════════════════
# TRAINING - YOLOv8l (LARGE) with ACCURACY-OPTIMIZED HYPERPARAMETERS
# 43.7M parameters - Higher accuracy than YOLOv8m
# ═══════════════════════════════════════════════════════════════════════════
from ultralytics import YOLO
import torch
from pathlib import Path

# Detect hardware
device = 0 if torch.cuda.is_available() else 'cpu'
is_gpu = torch.cuda.is_available()

print("="*80)
print(f"🖥️  Training Device: {'GPU - ' + torch.cuda.get_device_name(0) if is_gpu else 'CPU'}")
print("="*80)

# Get absolute path to SHARED dataset from yolo8m folder
data_yaml = Path.cwd().parent / "yolo8m" / "dataset" / "data.yaml"
print(f"Dataset config: {data_yaml}")
print(f"Exists: {data_yaml.exists()}\n")

# ═══════════════════════════════════════════════════════════════════════════
# AUTO-RESUME: Check if training was interrupted
# ═══════════════════════════════════════════════════════════════════════════
checkpoint_path = Path.cwd() / "runs" / "yolo8l_high_accuracy" / "weights" / "last.pt"
resume_training = checkpoint_path.exists()

if resume_training:
    print("🔄 RESUMING TRAINING from checkpoint")
    print(f"📁 Loading: {checkpoint_path}")
    model = YOLO(str(checkpoint_path))
    
    # Check which epoch we're at
    results_csv = Path.cwd() / "runs" / "yolo8m_high_accuracy" / "results.csv"
    if results_csv.exists():
        with open(results_csv, 'r') as f:
            lines = f.readlines()
            last_epoch = len(lines) - 2  # -2 because of header and 0-indexing
            print(f"📊 Last completed epoch: {last_epoch}")
            print(f"▶️  Will resume from epoch {last_epoch + 1}")
    print()
else:
    print("🆕 STARTING FRESH TRAINING")
    print("📁 Loading base model: yolov8l.pt (LARGE - 43.7M params)")
    model = YOLO('yolov8l.pt')
    print()

# ═══════════════════════════════════════════════════════════════════════════
# ACCURACY-OPTIMIZED HYPERPARAMETERS
# Based on best practices for exam surveillance detection
# ═══════════════════════════════════════════════════════════════════════════

# Hardware-adaptive settings for maximum accuracy
if is_gpu:
    # GPU: Maximum accuracy configuration
    EPOCHS = 150
    IMGSZ = 1280      # High resolution for better small object detection
    BATCH = 8         # Larger batch = more stable training
    WORKERS = 8
    CACHE = True      # Cache images in RAM for speed
    AMP = True        # Mixed precision for speed without accuracy loss
    PATIENCE = 30
    print("✓ Using GPU configuration for MAXIMUM ACCURACY")
else:
    # CPU: Balanced accuracy (still high quality, just slower)
    EPOCHS = 50      # More epochs for better learning
    IMGSZ = 832       # Medium-high resolution (compromise for CPU)
    BATCH = 2         # Small batch for CPU memory
    WORKERS = 2
    CACHE = False
    AMP = False
    PATIENCE = 25
    print("✓ Using CPU configuration for HIGH ACCURACY (slower training)")
    print(f"⏱️  Estimated training time: ~15-25 hours")

print(f"\nTraining Configuration:")
print(f"  • Epochs: {EPOCHS}")
print(f"  • Image Size: {IMGSZ}x{IMGSZ}")
print(f"  • Batch Size: {BATCH}")
print("="*80 + "\n")

results = model.train(
    # ═══════════════════════════════════════════════════════════════════════
    # DATASET
    # ═══════════════════════════════════════════════════════════════════════
    data=str(data_yaml),
    
    # ═══════════════════════════════════════════════════════════════════════
    # CORE SETTINGS - Optimized for accuracy
    # ═══════════════════════════════════════════════════════════════════════
    epochs=EPOCHS,
    imgsz=IMGSZ,
    batch=BATCH,
    device=device,
    cache=CACHE,
    
    # ═══════════════════════════════════════════════════════════════════════
    # OPTIMIZER - Optimized for YOLOv8l with CPU-friendly convergence
    # ═══════════════════════════════════════════════════════════════════════
    optimizer='AdamW',           # Best optimizer for YOLOv8
    lr0=0.0008,                  # Lower LR for stable training (large model)
    lrf=0.00008,                 # Very low final LR for fine details
    momentum=0.937,              # Standard momentum
    weight_decay=0.0006,         # Regularization (prevent overfitting)
    warmup_epochs=6.0,           # Longer warmup for smooth start
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    
    # ═══════════════════════════════════════════════════════════════════════
    # DATA AUGMENTATION - CPU-friendly & accuracy-optimized
    # These augmentations don't impact CPU performance significantly
    # ═══════════════════════════════════════════════════════════════════════
    # Color augmentation
    hsv_h=0.015,                 # Hue variation
    hsv_s=0.7,                   # Saturation
    hsv_v=0.4,                   # Brightness (exam lighting)
    
    # Geometric augmentation
    degrees=15.0,                # Rotation
    translate=0.1,               # Position shifts
    scale=0.6,                   # Scale variation
    shear=2.0,                   # Perspective
    perspective=0.0001,
    
    # Flip augmentation
    flipud=0.0,                  # No vertical flip
    fliplr=0.5,                  # Horizontal flip
    
    # Advanced augmentation (CRITICAL - minimal CPU overhead)
    mosaic=1.0,                  # Mosaic (learns multiple students)
    mixup=0.2,                   # Mixup (generalization)
    copy_paste=0.1,              # Copy-paste (rare behaviors)
    
    # ═══════════════════════════════════════════════════════════════════════
    # LOSS WEIGHTS - Optimized for YOLOv8l (larger model capacity)
    # ═══════════════════════════════════════════════════════════════════════
    box=7.5,                     # Bounding box loss
    cls=0.8,                     # HIGHER for large model (better class discrimination)
    dfl=1.6,                     # Slightly higher for precise localization
    
    # ═══════════════════════════════════════════════════════════════════════
    # TRAINING CONTROL - For maximum accuracy
    # ═══════════════════════════════════════════════════════════════════════
    patience=PATIENCE,           # Early stopping patience
    save_period=10,              # Save checkpoint every 10 epochs
    cos_lr=True,                 # Cosine LR schedule (smooth decay)
    close_mosaic=20,             # Disable mosaic in last 20 epochs (fine-tune on real images)
    amp=AMP,                     # Mixed precision training
    fraction=1.0,                # Use 100% of dataset
    
    # ═══════════════════════════════════════════════════════════════════════
    # VALIDATION & PERFORMANCE
    # ═══════════════════════════════════════════════════════════════════════
    val=True,                    # Validate during training
    workers=WORKERS,
    rect=False,                  # Don't use rectangular training (better augmentation)
    
    # ═══════════════════════════════════════════════════════════════════════
    # SAVING
    # ═══════════════════════════════════════════════════════════════════════
    project='runs',
    name='yolo8m_high_accuracy',
    save=True,
    exist_ok=True,
    plots=True,
    verbose=True,
    resume=True,
)

print("\n" + "="*80)
print("✓✓✓ YOLOv8l TRAINING COMPLETE! ✓✓✓")
print("="*80)
print(f"📁 Best model: runs/yolo8l_high_accuracy/weights/best.pt")
print(f"📁 Last model: runs/yolo8l_high_accuracy/weights/last.pt")
print(f"📊 Training plots: runs/yolo8l_high_accuracy/")
print(f"\n💡 YOLOv8l has 43.7M parameters (vs YOLOv8m's 25.9M)")
print(f"   Expected: +2-4% mAP improvement over YOLOv8m")

# Display final metrics
if hasattr(results, 'results_dict'):
    print(f"\n📈 Final Training Metrics:")
    print(f"  • mAP@50: {results.results_dict.get('metrics/mAP50(B)', 'N/A')}")
    print(f"  • mAP@50-95: {results.results_dict.get('metrics/mAP50-95(B)', 'N/A')}")

print("="*80)
```

```python
# ═══════════════════════════════════════════════════════════════════════════
# COMPREHENSIVE TEST SET EVALUATION
# ═══════════════════════════════════════════════════════════════════════════
from ultralytics import YOLO
import pandas as pd
from pathlib import Path
import torch

# Detect hardware
device = 0 if torch.cuda.is_available() else 'cpu'
is_gpu = torch.cuda.is_available()

# Load best model from training
model_path = Path.cwd() / "runs" / "yolo8l_high_accuracy" / "weights" / "best.pt"

if not model_path.exists():
    print(f"❌ Model not found at: {model_path}")
    print("Please complete training first!")
else:
    print("="*90)
    print(" "*25 + "📊 COMPREHENSIVE MODEL EVALUATION")
    print("="*90)
    
    model = YOLO(str(model_path))
    
        # Get absolute path to SHARED dataset
    data_yaml = Path.cwd().parent / "yolo8m" / "dataset" / "data.yaml"
    
    # Match evaluation settings to training
    eval_imgsz = 1280 if is_gpu else 832
    eval_batch = 8 if is_gpu else 2
    
    print(f"Evaluating on test set...")
    print(f"  • Image size: {eval_imgsz}x{eval_imgsz}")
    print(f"  • Batch size: {eval_batch}")
    print(f"  • Device: {'GPU' if is_gpu else 'CPU'}\n")
    
    # COMPREHENSIVE TEST EVALUATION
    test_results = model.val(
        data=str(data_yaml),
        split='test',
        imgsz=eval_imgsz,        # Match training resolution
        batch=eval_batch,
        device=device,
        plots=True,              # Generate confusion matrix, PR curves
        save_json=True,          # Save predictions in COCO format
        conf=0.20,               # Lower threshold for recall (0.20 for accuracy tests)
        iou=0.65,                # Higher IoU threshold for precision
        max_det=100,             # Max detections per image (exam scenarios)
        verbose=True
    )
    
    # ═══════════════════════════════════════════════════════════════════════
    # DISPLAY COMPREHENSIVE RESULTS
    # ═══════════════════════════════════════════════════════════════════════
    
    print("\n" + "="*90)
    print(" "*30 + "📈 OVERALL METRICS")
    print("="*90)
    
    mAP50 = test_results.box.map50 * 100
    mAP50_95 = test_results.box.map * 100
    precision = test_results.box.mp * 100
    recall = test_results.box.mr * 100
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    
    print(f"\n  🎯 mAP@50           : {mAP50:6.2f}%  {'🟢 Excellent' if mAP50 > 90 else '🟡 Good' if mAP50 > 80 else '🟠 Fair'}")
    print(f"  🎯 mAP@50-95        : {mAP50_95:6.2f}%  {'🟢 Excellent' if mAP50_95 > 70 else '🟡 Good' if mAP50_95 > 60 else '🟠 Fair'}")
    print(f"  ✓ Precision         : {precision:6.2f}%  (Low false positives)")
    print(f"  ✓ Recall            : {recall:6.2f}%  (Catches violations)")
    print(f"  ⚖️  F1-Score         : {f1:6.2f}%  (Overall balance)")
    
    print(f"\n" + "="*90)
    print(" "*30 + "⚡ INFERENCE SPEED")
    print("="*90)
    
    preprocess = test_results.speed['preprocess']
    inference = test_results.speed['inference']
    postprocess = test_results.speed['postprocess']
    total = sum(test_results.speed.values())
    fps = 1000 / total
    
    print(f"\n  Preprocess       : {preprocess:7.2f} ms")
    print(f"  Inference        : {inference:7.2f} ms  (core model)")
    print(f"  Postprocess      : {postprocess:7.2f} ms  (NMS)")
    print(f"  ─────────────────────────────")
    print(f"  Total/image      : {total:7.2f} ms")
    print(f"  Throughput       : {fps:7.1f} FPS  {'✓ Real-time' if fps > 30 else '⚠️ Near real-time' if fps > 15 else '⏱️ Offline'}")
    
    print(f"\n" + "="*90)
    print(" "*30 + "🎯 PER-CLASS PERFORMANCE")
    print("="*90)
    
    classes = ['Bend Over The Desk', 'Hand Under Table', 'Look Around',
               'Normal', 'phone', 'Stand Up', 'Wave']
    
    print(f"\n{'Class':<25} {'Precision':>11} {'Recall':>11} {'mAP50':>11} {'mAP50-95':>11} {'F1':>11}")
    print("─"*90)
    
    class_results = []
    for i, cls in enumerate(classes):
        if i < len(test_results.box.p):
            p = test_results.box.p[i] * 100
            r = test_results.box.r[i] * 100
            ap50 = test_results.box.ap50[i] * 100
            ap = test_results.box.ap[i] * 100
            f1_class = 2*(p*r)/(p+r) if (p+r) > 0 else 0
            
            print(f"{cls:<25} {p:>10.1f}% {r:>10.1f}% {ap50:>10.1f}% {ap:>10.1f}% {f1_class:>10.1f}%")
            
            class_results.append({
                'Class': cls,
                'Precision': p,
                'Recall': r,
                'mAP50': ap50,
                'mAP50-95': ap,
                'F1-Score': f1_class
            })
    
    print("\n" + "="*90)
    print(" "*30 + "📦 MODEL INFORMATION")
    print("="*90)
    print(f"\n  Architecture     : YOLOv8m (Medium)")
    print(f"  Parameters       : ~25.9M")
    print(f"  Input Size       : {eval_imgsz}x{eval_imgsz}")
    print(f"  Model Size       : ~52 MB")
    
    # Save comprehensive results
    results_dir = Path.cwd() / "results"
    
    # Save detailed CSV
    csv_path = results_dir / 'detailed_test_results.csv'
    results_df = pd.DataFrame(class_results)
    
    # Add overall metrics
    overall_row = pd.DataFrame([{
        'Class': 'OVERALL',
        'Precision': precision,
        'Recall': recall,
        'mAP50': mAP50,
        'mAP50-95': mAP50_95,
        'F1-Score': f1
    }])
    
    final_df = pd.concat([overall_row, results_df], ignore_index=True)
    final_df.to_csv(csv_path, index=False)
    
        # Save summary
    summary_path = results_dir / 'yolo8l_model_summary.txt'
    with open(summary_path, 'w') as f:
        f.write("="*90 + "\n")
        f.write("YOLOv8l (LARGE) EXAM SURVEILLANCE MODEL - TEST SET EVALUATION\n")
        f.write("="*90 + "\n\n")
        f.write(f"Overall Performance:\n")
        f.write(f"  mAP@50       : {mAP50:.2f}%\n")
        f.write(f"  mAP@50-95    : {mAP50_95:.2f}%\n")
        f.write(f"  Precision    : {precision:.2f}%\n")
        f.write(f"  Recall       : {recall:.2f}%\n")
        f.write(f"  F1-Score     : {f1:.2f}%\n")
        f.write(f"\nInference Speed:\n")
        f.write(f"  FPS          : {fps:.1f}\n")
        f.write(f"  Latency      : {total:.2f} ms/image\n")
        f.write(f"\nModel Size     : ~87 MB\n")
        f.write(f"Architecture   : YOLOv8l (43.7M parameters)\n")
        f.write(f"Input Size     : {eval_imgsz}x{eval_imgsz}\n")
    
    print(f"\n💾 Results saved:")
    print(f"  • {csv_path}")
    print(f"  • {summary_path}")
    print(f"  • Plots: runs/yolo8l_high_accuracy/")
    
    print("\n" + "="*90)
    print("✓✓✓ EVALUATION COMPLETE! ✓✓✓")
    print("="*90)
```
